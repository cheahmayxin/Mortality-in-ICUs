```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

ICU = pd.read_csv("Preprocessed ICU data.csv")
ICU.head()
```

```python
ICU.shape
```

```python
ICU.isna().sum()
```

```python
ICU.describe
```

```python
# Change Gender: Male = 1, Female = 0
ICU['Gender'] = ICU['Gender'].map({'male': 1, 'female': 0})
```

```python
ICU['Gender'].value_counts(dropna=False)
```

```python
# Check missing value percentage for each column
missing_percentage = ICU.isna().mean() * 100
missing_percentage.sort_values(ascending=False)
```

```python
# Find columns with more than 75% missing values
columns_above_75 = missing_percentage[missing_percentage > 75].index
columns_above_75.tolist()
```

```python
# Remove columns with more than 75% missing values
ICU = ICU.drop(columns=columns_above_75)
```

```python
# Check strange Height values
ICU[ICU['Height'] > 250]
```

```python
# Remove only unrealistic Height values, but keep missing values
ICU = ICU[(ICU['Height'] <= 250) | (ICU['Height'].isna())]
```

```python
ICU['Height'].describe()
```

```python
# Check duplicated rows
ICU.duplicated().sum()
```

```python
# Remove duplicated rows
ICU = ICU.drop_duplicates()
```

```python
# Check duplicated columns
duplicate_columns = ICU.T.duplicated()
ICU.columns[duplicate_columns].tolist()
```

```python
# Remove duplicated columns
ICU = ICU.loc[:, ~ICU.T.duplicated()]
```

```python
# Check which numerical columns contain decimal values
numeric_columns = ICU.select_dtypes(include='number')

decimal_columns = [
    col for col in numeric_columns.columns
    if ((numeric_columns[col].dropna() % 1) != 0).any()
]

decimal_columns
```

```python
ICU = ICU.round(4)
```
